```python
df25 = df25[df25['MainBranch'].isin(['I am a developer by profession', 'I am not primarily a developer, but I write code sometimes as part of my work/studies'])]
df25 = df25[(df25['ConvertedCompYearly']>=100) & (df25['ConvertedCompYearly']<1000000)]
df25
```

```python
edu_grp25 = df25.groupby(['EdLevel'])
edu_grp25_cleaned = edu_grp25['ConvertedCompYearly'].apply(lambda x: x[((x>=(x.quantile(0.25) - 1.5*(x.quantile(0.75)-x.quantile(0.25)))) & (x<=(x.quantile(0.75) + 1.5*(x.quantile(0.75)-x.quantile(0.25)))))])
edu_grp25_cleaned
```

```python
ed_mapping = {
    'Bachelor’s degree (B.A., B.S., B.Eng., etc.)': 'Undergraduate',
    'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)': 'Master’s Degree',
    'Some college/university study without earning a degree': 'Some College without Degree',
    'Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)': 'High School',
    'Associate degree (A.A., A.S., etc.)': 'Associate Degree',
    'Professional degree (JD, MD, Ph.D, Ed.D, etc.)': 'PhD/Professional',
    'Professional degree (JD, MD, etc.)': 'PhD/Professional', 
    'Other doctoral degree (Ph.D., Ed.D., etc.)': 'PhD/Professional',
    'Primary/elementary school': 'Primary School',
    'Something else': 'Other', 
    'Other (please specify):': 'Other'  
}
```

```python
edu_grp25_cleaned_graph = edu_grp25_cleaned.groupby('EdLevel').agg(['median']).sort_values(by='median', ascending=False)
edu_grp25_cleaned_graph.index = edu_grp25_cleaned_graph.index.map(ed_mapping)
plt.barh(edu_grp25_cleaned_graph.index, edu_grp25_cleaned_graph['median'], label='Median Salary')
plt.xticks(rotation=45)
plt.xlabel('Salary (USD)')
plt.xlim(0,90000)    
plt.title('Median Salary by Education Level (2025)')
```

```python
m23 = edu_grp23_cleaned_graph
m24 = edu_grp24_cleaned_graph
m25 = edu_grp25_cleaned_graph
comparison_edu = pd.concat([m23[m23.index.isin(['Undergraduate', 'Master’s Degree', 'PhD/Professional', 'High School'])]['median'],
                            m24[m24.index.isin(['Undergraduate', 'Master’s Degree', 'PhD/Professional', 'High School'])]['median'], 
                            m25[m25.index.isin(['Undergraduate', 'Master’s Degree', 'PhD/Professional', 'High School'])]['median']],
                            axis=1)
comparison_edu.columns = ['2023', '2024', '2025']
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))
comparison_edu.plot(kind='bar', ax=ax1, width=0.8, color=['#1f77b4', '#ff7f0e', '#2ca02c'])
ax1.set_xlabel('Education Level')
ax1.set_ylabel('Median Salary (USD)')
ax1.legend(title='Year')
ax2 = ax1.twinx()
ax2.plot(growth_df.index, growth_df['Total_Growth (%)'], marker='o', label='Growth 2024-2025', color='RED',alpha=0.7)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
new_labels = ['2023', '2024', '2025', 'Salary Growth % (2023-2025)']
ax1.legend(lines1 + lines2, new_labels, frameon=True, shadow=True, bbox_to_anchor=(0.5, 1.15),)
ax2.tick_params(axis='y', labelcolor='RED')
ax1.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)
ax2.grid(False)
```

Now the third question, the relation between the Education level and resources used to learn.

```python
relation = df25_copy.copy()
relation['EdLevel'] = relation['EdLevel'].map(ed_mapping)
relation['LearnCodeSeparated'] = relation['LearnCode'].str.split(';')
relation['EdLevel'].dropna(inplace=True)
relation['LearnCodeSeparated'].dropna(inplace=True)
```

```python
relation_exploded = relation[['EdLevel', 'LearnCodeSeparated']].explode('LearnCodeSeparated')
```

```python
relation_exploded.dropna(subset=['EdLevel', 'LearnCodeSeparated'], inplace=True, how='any')
```

```python
import seaborn as sns
```

```python
relation_exploded['LearnCodeSeparated'] = relation_exploded['LearnCodeSeparated'].map(lambda x: sources_mapping.get(x, x))
heatmap_data = relation_exploded.pivot_table(index='EdLevel', columns='LearnCodeSeparated', aggfunc='size', fill_value=0)
plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, annot=True, fmt='d', cmap='YlGnBu', )
plt.title('Learning Resources by Education Level (2025)', fontsize=16, fontweight='bold')
plt.xlabel('Learning Resource', fontsize=14, fontweight='bold')
plt.ylabel('Education Level', fontsize=14, fontweight='bold')
plt.show()
```

```python
a = relation_exploded.groupby(['EdLevel', 'LearnCodeSeparated']).size().unstack(fill_value=0)
a.rename(columns={'Other online resources (e.g. standard search, forum, online community)': 'Other Online Resources', 'Other (please specify):': 'Other Resources'}, inplace=True)
a
```

```python
heatmap_data2 = a.apply(lambda x: (x / x.sum()) * 100, axis=1)
heatmap_data2
```

```python
plt.figure(figsize=(16, 10))
labels = heatmap_data2.applymap(lambda v: f'{v:.1f}%').values
sns.heatmap(heatmap_data2, annot=labels, fmt='', cmap='YlGnBu', )
plt.title('Learning Resource Preference by Education Level (2025) [%]', fontsize=18, fontweight='bold')
plt.xlabel('Learning Resource', fontsize=16, fontweight='bold')
plt.ylabel('Education Level', fontsize=16, fontweight='bold')
plt.show()
```

Now we will go to the next section which is the relation between DevType and different education levels.

```python
selected_jobs = [
    'Academic researcher', 'Applied scientist', 'AI/ML engineer', 
    'Data scientist', 'Developer, full-stack', 'Developer, back-end', 
    'Developer, front-end', 'Developer, mobile', 'Engineering manager', 
    'DevOps engineer or professional','Data or business analyst'
]
```

```python
dev_relation = df25_copy[['EdLevel', 'DevType']]
dev_relation['EdLevel'] = dev_relation['EdLevel'].map(ed_mapping)
dev_relation.dropna(subset=['EdLevel', 'DevType'], inplace=True, how='any')
dev_transposed = dev_relation.groupby(['EdLevel', 'DevType']).size().unstack(fill_value=0).transpose()
dev_pct = dev_transposed.div(dev_transposed.sum(axis=0), axis=1) * 100
dev_pct = dev_pct[dev_pct.index.isin(selected_jobs)]
dev_pct = dev_pct.drop(columns=['Other','Primary School'], errors='ignore')
dev_pct
```

```python
plt.figure(figsize=(16, 10))
labels = dev_pct.applymap(lambda v: f'{v:.1f}%').values
sns.heatmap(dev_pct, annot=labels, fmt='', cmap='YlGnBu')
plt.xlabel('Education Level', fontsize=14, fontweight='bold')
plt.ylabel('Developer Type', fontsize=14, fontweight='bold')
plt.title('Developer Role Distribution by Education Level (2025)', fontsize=16, fontweight='bold')
plt.text(0.25, -0.25, 
         'Note: Percentages are normalized per education group to show relative developer roles preference.', 
         ha='center', va='center', transform=plt.gca().transAxes, 
         fontsize=11, color='#555555', style='italic')
plt.show()
```

Now the next question is the relation between the Languages and the education level

```python
lan_relation = df25_copy[['LanguageHaveWorkedWith', 'EdLevel']]
lan_relation['EdLevel'] = lan_relation['EdLevel'].map(ed_mapping)
lan_relation.dropna(subset=['EdLevel', 'LanguageHaveWorkedWith'], inplace=True, how='any')
lan_relation['LanguageHaveWorkedWith'] = lan_relation['LanguageHaveWorkedWith'].str.split(';')
lan_relation_exploded = lan_relation.explode('LanguageHaveWorkedWith')
lan_grp = lan_relation_exploded.groupby(['EdLevel', 'LanguageHaveWorkedWith']).size().unstack(fill_value=0)
lan_transpoed = lan_grp.transpose()
lang_pct = lan_transpoed.div(lan_transpoed.sum(axis=0), axis=1) * 100
lang_pct = lang_pct[lang_pct.index.isin(['Python', 'JavaScript', 'SQL', 'HTML/CSS', 'R','Go','Rust','TypeScript','C','C++','C#','Java','PHP'])]
lang_pct
```

```python
plt.figure(figsize=(16, 10))
labels = lang_pct.applymap(lambda v: f'{v:.1f}%').values
sns.heatmap(lang_pct, annot=labels, fmt='', cmap='YlGnBu')
plt.xlabel('Education Level', fontsize=14, fontweight='bold')
plt.ylabel('Programming Language', fontsize=14, fontweight='bold')
plt.title('Programming Language Market Share within Each Education Level (2025)', fontsize=16, fontweight='bold')
plt.text(0.25, -0.25, 
         'Note: Percentages are normalized per education group to show relative language preference.', 
         ha='center', va='center', transform=plt.gca().transAxes, 
         fontsize=11, color='#555555', style='italic')
plt.show()
```
